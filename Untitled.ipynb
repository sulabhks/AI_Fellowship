### Write a program to calculate the factorial of a number using FOR loop

```python
number = int(input("Enter a nmber"))

fact=1
for i in range(1, number+1):
    fact*=i
print(f"{number}!= {fact}")
```

### Write a program to calculate the factorial of a number using function

```python
number=int(input("Enter a number"))

def factorial(number):
    fact=1
    for i in range(1,number+1):
        fact*=i
    return fact

print(f"{number}!={factorial(number)}")
```

### Write a program to calculate the factorial of a number using recursive function.

```python
number=int(input("Enter a number"))

def factorial(n):
    if n<0:
        print(f"{n} is negative. Factorial does not exist.")
    elif n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)

print(f"{number}!={factorial(number)}")
```
