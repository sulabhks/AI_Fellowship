### Exercise on Arithmetic Operators

```python
# 1. Addition 
number1=int(input("Enter the first number"))
number2=int(input("Enter the second number"))

Results=number1+number2
print(f"Addition of {number1} + {number2} = {Results}")
```

```python
# 2. Substraction
number1=int(input("Enter the first number"))
number2=int(input("Enter the second number"))

Results=number1-number2
print(f"Substraction of {number1} - {number2} = {Results}")
```

```python
# 3. Multiplication
number1=int(input("Enter the first number"))
number2=int(input("Enter the second number"))

Results=number1*number2
print(f"Multiplication of {number1} * {number2} = {Results}")
```

```python
# 4. Division
number1=int(input("Enter the first number"))
number2=int(input("Enter the second number"))

Results=number1/number2
print(f"Division of {number1} / {number2} = {Results}")
```

```python
# 5. Exponentitation
number1=int(input("Enter the first number"))
number2=int(input("Enter the second number"))

Results=number1**number2
print(f"Exponentiation of {number1} ** {number2} = {Results}")
```

```python
# 6. Floor Division
number1=int(input("Enter the first number"))
number2=int(input("Enter the second number"))

Results=number1//number2
print(f"Floor Division of {number1} // {number2} = {Results}")
```

```python
# 7. Modulus
number1=int(input("Enter the first number"))
number2=int(input("Enter the second number"))

Results=number1%number2
print(f"Modulous of {number1} % {number2} = {Results}")
```

### Exercise on Assignment Operators

```python
# 1. Simple Assignment
var=int(input("Enter a number"))
print(var)
type(var)
```

```python
# 2. Add and Assign
var=int(input("Enter a number"))
var+=5
print(var)
```

```python
# 3. Subtract and Assign
var=int(input("Enter a number"))
var-=5
print(var)
```

```python
# 4. Multiply and Assign
var=int(input("Enter a number"))
var*=5
print(var)
```

```python
# 5. Divide and Assign
var=int(input("Enter a number"))
var/=5
print(var)
```

```python
# 6. Modulus and Assign
var=int(input("Enter a number"))
var%=5
print(var)
```

```python
# 7. Floor Division and Assign
var=int(input("Enter a number"))
var//=5
print(var)
```

```python
# 8. Exponent and Assign
var=int(input("Enter a number"))
var**=5
print(var)
```

```python
# 9. Bitwise AND and Assign
var=int(input("Enter a number"))
var&=5
print(var)
```

```python
# 10. Bitwise OR and Assign
var=int(input("Enter a number"))
var|=5
print(var)
```

```python
# 11. Bitwise XOR and Assign
var=int(input("Enter a number"))
var^=5
print(var)
```

```python
# 11. Right shift and Assign
var=int(input("Enter a number"))
var>>=2
print(var)
```

```python
# 12. Left shift and Assign
var=int(input("Enter a number"))
var<<=3
print(var)
```

```python
# 13. Walrus Operator (Assignment Expression)
# var=3
# print(var)
print(var:=3)
```
